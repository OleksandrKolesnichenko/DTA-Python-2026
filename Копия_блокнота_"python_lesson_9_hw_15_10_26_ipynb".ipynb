## Завдання 1. Перевірка та очищення даних

Переглянь усі поїздки та розподіли їх:
- скасовані поїздки;
- **некоректні** завершені поїздки: відстань дорівнює 0 або менше — причина `"invalid distance"`; оцінка не є цілим числом від 1 до 5 — причина `"invalid rating"`. Якщо в поїздці кілька проблем, вказуй лише першу з них у цьому порядку;
- **коректні** завершені поїздки — збережи їх в окремий список `valid_rides`, з ним ти працюватимеш далі.

Виведи загальну кількість поїздок, кількість скасованих, кількість коректних, а також номер і причину для кожної некоректної поїздки

```python
cancelled_rides = []
incorrect_rides = []
valid_rides = []
for ride in rides:
  if ride["status"] == "cancelled":
    cancelled_rides.append(ride)
  elif ride['status'] == 'completed':
    if ride['distance'] <= 0:
      incorrect_rides.append({"id": ride["id"], "reason": "Invalid distance"})
    elif not isinstance(ride['rating'], int) or not (1 <= ride['rating'] <= 5):
      incorrect_rides.append({"id": ride["id"], "reason": "Invalid rating"})
    else: valid_rides.append(ride)

print(f"Загальна кількість поїздок: {len(rides)}")
print(f"Кількість скасованих поїздок: {len(cancelled_rides)}")
print(f"Кількість коректних завершених поїздок: {len(valid_rides)}")
print()
print("Некоректні поїздки:")

for item in incorrect_rides:
    print(f"ID: {item['id']}, Причина: {item['reason']}")
```

## Завдання 2. Розрахунок вартості поїздок

Для кожної поїздки з `valid_rides` розрахуй вартість за тарифом:
- посадка — 50 грн плюс 12 грн за кожен кілометр;
- якщо поїздка тривала понад 30 хвилин — додатково 3 грн за кожну хвилину понад 30;
- нічний тариф (поїздка почалася з 22:00 до 05:59 включно) — уся вартість збільшується на 25%;
- мінімальна вартість поїздки — 80 грн: якщо розрахована сума менша, пасажир платить 80 грн.

Округли вартість до 2 знаків і додай її до словника кожної поїздки з ключем `"price"`. Виведи для кожної поїздки номер, водія, відстань і вартість

```python

import pandas as pd
df_rides = pd.DataFrame(rides)

df_rides.head()
```

```python
for ride in rides:
  base_price = 50+12 * ride['distance']
  if ride['duration'] > 30:
    base_price += (ride['duration'] - 30) * 3
  if ride['hour'] >= 22 or ride['hour'] <= 5:
    base_price *=1.25
  if base_price < 80:
    base_price = 80
  ride['price'] = round(base_price, 2)
print(f"{'ID':<6} | {'Водій':<10} | {'Відстань':<10} | {'Вартість':<10}")
print()

for ride in valid_rides:
  print(f"{ride['id']:<6} | {ride['driver']:<10} | {ride['distance']:<10} | {ride['price']:<10}")
```

## Завдання 3. Статистика по водіях

На основі `valid_rides` створи словник `driver_stats`, у якому для кожного водія зберігаються:
- кількість поїздок;
- виручка (сума вартостей його поїздок, округлена до 2 знаків);
- середній рейтинг, округлений до 1 знака.

Виведи статистику кожного водія в одному рядку

```python
driver_stats = {}

for ride in valid_rides:
  driver = ride['driver']
  if driver not in driver_stats:
    driver_stats[driver] = {
      "rides_count": 0,
      "total_revenue": 0.0,
      "ratings_sum": 0.0
    }
  driver_stats[driver]["rides_count"] += 1
  driver_stats[driver]["total_revenue"] += ride["price"]
  driver_stats[driver]['ratings_sum'] += ride['rating']
for driver, stats in driver_stats.items():
  stats["total_revenue"] = round(stats["total_revenue"], 2)
  avg_rating = stats["ratings_sum"] / stats["rides_count"]
  stats["avg_rating"] = round(avg_rating, 1)
  del stats["ratings_sum"]
print(f"{'Водій':<10} | {'Кількість поїздок':<18} | {'Виручка':<10} | {'Середній рейтинг':<16}")
print()
for driver, stats in driver_stats.items():

 print(f"{driver:<10} | {stats['rides_count']:<18} | {stats['total_revenue']:<10} | {stats['avg_rating']:<16}")
```

## Завдання 4. Статуси та бонуси водіїв

На основі `driver_stats` визнач для кожного водія статус і бонус (середній рейтинг береться округленим до 1 знака):
- рейтинг від 4.8 **і** щонайменше 3 поїздки — статус `"Gold"`, бонус 10% від виручки;
- рейтинг від 4.5 — статус `"Silver"`, бонус 5% від виручки;
- рейтинг нижче 4.0 — статус `"Warning"`, бонусу немає;
- в інших випадках — статус `"Standard"`, бонусу немає.

Округли бонус до 2 знаків і додай статус та бонус до статистики водія. Виведи імʼя, статус і бонус кожного водія, а наприкінці — загальну суму бонусів

```python
total_bonuses = 0.0
for driver, stats in driver_stats.items():
  rides_count = stats["rides_count"]
  avg_rating = stats["avg_rating"]
  revenue = stats["total_revenue"]
  if avg_rating >= 4.8 and rides_count >= 3:
    status = "Gold"
    bonus = revenue * 0.10
  elif avg_rating >= 4.5:
    status = "Silver"
    bonus = revenue * 0.05
  elif avg_rating <= 4.0:
    status = "Warning"
    bonus = 0.0
  else:
    status = "Standard"
    bonus = 0.0
  bonus = round(bonus, 2)
  stats["status"] = status
  stats["bonus"] = bonus
  total_bonuses += bonus
total_bonuses = round(total_bonuses, 2)

print(f"{'Водій':<10} | {'Статус':<10} | {'Бонус':<10}")
print()
for driver, stats in driver_stats.items():
    print(f"{driver:<10} | {stats['status']:<10} | {stats['bonus']:<10}")
print()
print(f"Загальна сума бонусів: {total_bonuses}")
```

## Завдання 5. Підсумковий звіт за день

Виведи звіт, у якому кожен показник займає окремий рядок:
- загальна виручка за день;
- кількість коректних поїздок і середня вартість поїздки (округли до 2 знаків);
- частка скасованих поїздок від усіх поїздок у журналі, % (округли до 1 знака);
- частка поїздок, оплачених карткою, серед коректних поїздок, % (округли до 1 знака);
- найзавантаженіша година — година, на яку припало найбільше коректних поїздок, і кількість поїздок у цю годину;
- найдовша поїздка: номер, водій і відстань;
- водій з найбільшою виручкою;
- загальна сума бонусів водіям.

Не використовуй функції `max()` і `min()`

```python

total_revenue_day = 0.0
for stats in driver_stats.values():
    total_revenue_day += stats["total_revenue"]
total_revenue_day = round(total_revenue_day, 2)


total_valid_rides = len(valid_rides)
total_revenue_for_avg = 0.0
for ride in valid_rides:
    total_revenue_for_avg += ride["price"]

if total_valid_rides > 0:
    avg_ride_price = round(total_revenue_for_avg / total_valid_rides, 2)
else:
    avg_ride_price = 0.0


cancelled_count = len(cancelled_rides)
total_rides_count = len(rides)
if total_rides_count > 0:
    cancelled_share = round((cancelled_count / total_rides_count) * 100, 1)
else:
    cancelled_share = 0.0


card_rides_count = 0
for ride in valid_rides:
    if ride["payment"] == "card":
        card_rides_count += 1

if total_valid_rides > 0:
    card_share = round((card_rides_count / total_valid_rides) * 100, 1)
else:
    card_share = 0.0

hours_count = {}
for ride in valid_rides:
    h = ride["hour"]
    if h not in hours_count:
        hours_count[h] = 0
    hours_count[h] += 1


busiest_hour = None
max_rides_in_hour = -1
for h, count in hours_count.items():
    if count > max_rides_in_hour:
        max_rides_in_hour = count
        busiest_hour = h


longest_ride_id = None
longest_ride_driver = None
max_distance = -1.0
for ride in valid_rides:
    if ride["distance"] > max_distance:
        max_distance = ride["distance"]
        longest_ride_id = ride["id"]
        longest_ride_driver = ride["driver"]


top_driver_name = None
max_driver_revenue = -1.0
for driver, stats in driver_stats.items():
    if stats["total_revenue"] > max_driver_revenue:
        max_driver_revenue = stats["total_revenue"]
        top_driver_name = driver


calculated_total_bonuses = 0.0
for stats in driver_stats.values():
    calculated_total_bonuses += stats["bonus"]
calculated_total_bonuses = round(calculated_total_bonuses, 2)

print(f"Загальна виручка за день: {total_revenue_day} грн")
print(f"Кількість коректних поїздок: {total_valid_rides}, середня вартість поїздки: {avg_ride_price} грн")
print(f"Частка скасованих поїздок: {cancelled_share}%")
print(f"Частка поїздок, оплачених карткою: {card_share}%")
print(f"Найзавантаженіша година: {busiest_hour}:00 ({max_rides_in_hour} поїздок)")
print(f"Найдовша поїздка: ID {longest_ride_id}, водій {longest_ride_driver}, відстань {max_distance} км")
print(f"Водій з найбільшою виручкою: {top_driver_name} ({max_driver_revenue} грн)")
print(f"Загальна сума бонусів водіям: {calculated_total_bonuses} грн")
```
