1. Відсортуй курси за кількістю студентів — від найпопулярнішого до найменш популярного — і виведи назву та кількість студентів кожного курсу.

2. Відбери курси з категорії `"data"` з рейтингом 4.5 і вище. Для відібраних курсів отримай:
- кількість таких курсів;
- список їхніх назв великими літерами;

Розвʼяжи завдання за допомогою `filter` і `map`

```python
filtered_courses =   list(filter(lambda x: x['category'] == 'data' and x['rating'] >= 4.5, COURSES))
count_courses = len(filtered_courses)
titl_up = list(map(lambda x: x["title"].upper(), filtered_courses))
price_curses_diskounted = list(map(lambda x: round(x["price"]*0.85), filtered_courses))
print(f'totsl top curses: {count_courses}')
print(f"Top curses: {titl_up}")
print(f"Ціни зі знижкою 15%: {price_curses_diskounted}")
```

3. Напиши функцію `top_by(items, field, n=3, condition=None)` з docstring, яка повертає `n` елементів з найбільшим значенням поля `field`. Якщо передано `condition` — функцію-умову, — спершу залиш лише ті елементи, для яких вона повертає `True`.

Виклич функцію й виведи назви курсів:
- 3 курси з найвищим рейтингом;
- 2 найпопулярніші курси (за кількістю студентів) серед курсів категорії `"bi"`;
- 1 найдовший курс (за кількістю годин)

```python
def top_by(items, field, n=3, condition=None):
  if condition is not None:
      items = list(filter(condition, items))
  sorted_items = sorted(items, key=lambda x: x[field], reverse=True)
  return sorted_items[:n]

top_rating = top_by(COURSES, field="rating", n=3)
print("1. TOP- 3 :")
for course in top_rating:
    print(f"  {course['title']} (рейтинг: {course['rating']})")
print()
top_bi_students = top_by(COURSES, field="students", n=2, condition=lambda x: x["category"] == "bi")
print("2. Топ-2 популярних курси в категорії 'bi':")
for course in top_bi_students:
    print(f"  {course['title']} (студентів: {course['students']})")
print()
longest_course = top_by(COURSES, field="hours", n=1)
print("3. Найдовший курс:")
for course in longest_course:
    print(f"  {course['title']} (годин: {course['hours']})")
```

4. Служба доставки бере 35 грн за кожен **розпочатий** кілограм ваги посилки: посилка вагою 1.2 кг оплачується як 2 кг.  
Дано ваги посилок у кілограмах: `weights = [0.4, 2.0, 3.7, 1.2, 12.5, 0.9]`.

Виведи вартість доставки кожної посилки та загальну вартість. Окремо виведи загальну вагу всіх посилок, округлену **вниз** до цілих кілограмів

```python
import math
weights = [0.4, 2.0, 3.7, 1.2, 12.5, 0.9]
total_cost = 0
for weight in weights:
    paid_kg = math.ceil(weight)
    cost = paid_kg*35
    total_cost +=cost
    print(f"реальна вага: {weight} kg  Оплачується: {paid_kg} kg  price: {cost} ")
print()
print(f"вартість доставки: {total_cost} грн")
print()
total_w = sum(weights)
rounded_total_w = math.floor(total_w)
print(f" загальну вагу всіх посилок {total_w}kg ")

print(f"Загальна вага всіх посилок (округлена вниз): {rounded_total_w} кг")

```

5. Дано час відповіді служби підтримки на звернення клієнтів (у хвилинах) за два тижні:

`week_1 = [12, 15, 9, 15, 22, 18, 15, 11, 14, 16]`  
`week_2 = [5, 30, 8, 27, 8, 25, 9, 31, 8, 28]`

За допомогою модуля `statistics` для кожного тижня виведи середнє, медіану, моду та стандартне відхилення (середнє й стандартне відхилення округли до 2 знаків).  
Виведи, у який тиждень служба працювала **стабільніше** (час відповіді менше коливався), і поясни в коментарі, на якому показнику ґрунтується висновок

```python
import statistics
week_1 = [12, 15, 9, 15, 22, 18, 15, 11, 14, 16]
week_2 = [5, 30, 8, 27, 8, 25, 9, 31, 8, 28]

def analyze_week(week_data, week_name):
    mean_val = round(statistics.mean(week_data), 2)
    median_val = statistics.median(week_data)
    mode_val = statistics.mode(week_data)
    stdev_val = round(statistics.stdev(week_data), 2)
    print(f"Статистика для {week_name}")
    print(f"Середнє: {mean_val}")
    print(f"Медіана: {median_val}")
    print(f"Мода: {mode_val}")
    print(f"Стандартне відхилення: {stdev_val}\n")
    return stdev_val
stdev_1 = analyze_week(week_1, "week 1")
stdev_2 = analyze_week(week_2, "week 2")
if stdev_1 < stdev_2:
    print("Висновок: week 1 служба працювала стабільніше.")
else:
    print("Висновок: week 2 служба працювала стабільніше.")
```

6. Згенеруй тестові дані для 10 замовлень. Встанови `seed` рівним 2026. Для кожного замовлення спочатку згенеруй суму — ціле число від 100 до 5000 включно, — а потім випадково обери спосіб оплати зі списку `["card", "cash", "apple pay"]`. Збережи замовлення як список кортежів `(сума, спосіб оплати)`.

Виведи:
- усі згенеровані замовлення;
- кількість замовлень кожним способом оплати;
- частку замовлень на суму понад 2000 грн у відсотках;
- 2 різних випадкових замовлення для перевірки якості, вибрані з уже згенерованого списку

```python
import random
random.seed(2345)
pay_methods = ["card", "cash", "apple pay"]

orders =[]
for i in range(10):
  amount = random.randint(10, 5000)
  method = random.choice(pay_methods)
  orders.append((amount, method))


for order in orders:
  print(f"CYMMA: {order[0]}грн, Price: {order[1]}")
print()

card_count = sum(1 for order in orders if order[1] == "card")
cash_count = sum(1 for order in orders if order[1] == 'cash')
app_count = sum(1 for order in orders if order[1] == 'apple pay')

print(f'totsl card pay count: {card_count}')
print(f'totsl cash pay count: {cash_count}')
print(f'totsl apple pay count: {app_count}')

exp_orderes = [order for order in orders if order[0] > 2000]
pct = (len(exp_orderes) / len(orders)) * 100
print(f"замовлень на суму понад 2000 грн у відсотках: {pct}")

chek = random.sample(orders, 2)
for order in chek:
  print(f" CYMMA: {order[0]}грн, Price: {order[1]}")
```

7. **Аналіз підписок.** Дано підписки клієнтів у форматі `(клієнт, дата початку, тривалість у днях)` і дату, на яку робиться звіт:

`TODAY = "2026-10-21"`  
`subscriptions = [("anna", "2026-09-01", 30), ("bohdan", "2026-10-01", 30), ("iryna", "2026-08-15", 90), ("oleh", "2026-10-10", 14), ("sofia", "2026-07-01", 365), ("taras", "2026-10-15", 7)]`

Для кожної підписки визнач:
- дату закінчення у форматі `ДД.ММ.РРРР`;
- день тижня, на який припадає дата закінчення (англійською, наприклад `Saturday`);
- скільки днів залишилося до закінчення від `TODAY` (для завершених підписок — відʼємне число);
- статус: `"expired"` — підписка вже закінчилася, `"expires soon"` — залишилося 7 днів або менше, `"active"` — в інших випадках.

Виведи результат таблицею за допомогою пакета `tabulate`. Наприкінці виведи кількість підписок кожного статусу та медіану кількості днів, що залишилися, для активних підписок

```python
from datetime import datetime, timedelta
import statistics
from tabulate import tabulate

TODAY = "2026-10-21"
subscriptions = [
    ("anna", "2026-09-01", 30),
    ("bohdan", "2026-10-01", 30),
    ("iryna", "2026-08-15", 90),
    ("oleh", "2026-10-10", 14),
    ("sofia", "2026-07-01", 365),
    ("taras", "2026-10-15", 7)
]
today_date = datetime.strptime(TODAY, "%Y-%m-%d")
table_data = []
status_counts = {"expired": 0, "expires soon": 0, "active": 0}
active_days_left = []
for client, start_str, duration in subscriptions:
    start_date = datetime.strptime(start_str, "%Y-%m-%d")
    end_date = start_date + timedelta(days=duration)
    end_date_str = end_date.strftime("%d.%m.%Y")
    day_of_week = end_date.strftime("%A")
    days_left = (end_date - today_date).days
    if days_left < 0:
        status = "expired"
    elif days_left <= 7:
        status = "expires soon"
    else:
        status = "active"
        active_days_left.append(days_left)
    status_counts[status] += 1
    table_data.append([client, end_date_str, day_of_week, days_left, status])
headers = ["Клієнт", "Дата закінчення", "День тижня", "Залишилось днів", "Статус"]

print()
print(tabulate(table_data, headers=headers, tablefmt="grid"))
print("Кільк підписок за статусами:")
print(f" expired: {status_counts['expired']}")
print(f"  expires soon: {status_counts['expires soon']}")
print(f"  active: {status_counts['active']}")
if active_days_left:
    median_active = statistics.median(active_days_left)
    print(f"Медіана кількості днів, що залишилися (для активних): {median_active}")
else:
    print("Немає активних підписок для розрахунку медіани.")
```
