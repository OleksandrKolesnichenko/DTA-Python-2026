## Завдання 1. Перевірка та очищення даних

Переглянь усі поїздки та розподіли їх:
- скасовані поїздки;
- **некоректні** завершені поїздки: відстань дорівнює 0 або менше — причина `"invalid distance"`; оцінка не є цілим числом від 1 до 5 — причина `"invalid rating"`. Якщо в поїздці кілька проблем, вказуй лише першу з них у цьому порядку;
- **коректні** завершені поїздки — збережи їх в окремий список `valid_rides`, з ним ти працюватимеш далі.

Виведи загальну кількість поїздок, кількість скасованих, кількість коректних, а також номер і причину для кожної некоректної поїздки

```python
cancelled_rides = []
invalid_rides = []
valid_rides = []
for ride in rides:
    if ride["status"] == "cancelled":
        cancelled_rides.append(ride)
    elif ride["status"] == "completed":
        if ride["distance"] <= 0:
            invalid_rides.append((ride["id"], "invalid distance"))
        elif not isinstance(ride["rating"], int) or isinstance(ride["rating"], bool) or not (1 <= ride["rating"] <= 5):
            invalid_rides.append((ride["id"], "invalid rating"))
        else:
            valid_rides.append(ride)
print(f"Кількість поїздок: {len(rides)}")
print(f"Кількість скасованих поїздок: {len(cancelled_rides)}")
print(f"Кількість коректних завершених поїздок (`valid_rides`): {len(valid_rides)}")
print(f"Кількість некоректних поїздок: {len(invalid_rides)}\n")
print("Список некоректних поїздок:")
for ride_id, reason in invalid_rides:
    print(f"Номер поїздки (id): {ride_id} — Причина: {reason}")
```

## Завдання 2. Розрахунок вартості поїздок

Для кожної поїздки з `valid_rides` розрахуй вартість за тарифом:
- посадка — 50 грн плюс 12 грн за кожен кілометр;
- якщо поїздка тривала понад 30 хвилин — додатково 3 грн за кожну хвилину понад 30;
- нічний тариф (поїздка почалася з 22:00 до 05:59 включно) — уся вартість збільшується на 25%;
- мінімальна вартість поїздки — 80 грн: якщо розрахована сума менша, пасажир платить 80 грн.

Округли вартість до 2 знаків і додай її до словника кожної поїздки з ключем `"price"`. Виведи для кожної поїздки номер, водія, відстань і вартість

```python
import pandas as pd
for ride in valid_rides:
    cost = 50 + 12 * ride['distance']

    if ride["duration"] > 30:
        cost += 3 * (ride['duration'] - 30)
    if ride['hour'] >= 22 or ride['hour'] <= 5:
        cost *= 1.25
    if cost < 80:
        cost = 80
    ride["price"] = round(cost, 2)
df_rides = pd.DataFrame(valid_rides)[["id", "driver", "distance", "price"]]
df_rides.columns = ["Номер поїздки", "Водій", "Відстань (км)", "Вартість (грн)"]
display(df_rides)
```

## Завдання 3. Статистика по водіях

На основі `valid_rides` створи словник `driver_stats`, у якому для кожного водія зберігаються:
- кількість поїздок;
- виручка (сума вартостей його поїздок, округлена до 2 знаків);
- середній рейтинг, округлений до 1 знака.

Виведи статистику кожного водія в одному рядку

```python
driver_stats = {}

for ride in valid_rides:
    driver = ride["driver"]
    price = ride["price"]
    rating = ride["rating"]

    if driver not in driver_stats:
        driver_stats[driver] = {
            "rides_count": 0,
            "total_revenue": 0.0,
            "ratings_sum": 0
        }
    driver_stats[driver]["rides_count"] += 1
    driver_stats[driver]["total_revenue"] += price
    driver_stats[driver]["ratings_sum"] += rating
for driver, stats in driver_stats.items():
    count = stats["rides_count"]
    revenue = round(stats["total_revenue"], 2)
    avg_rating = round(stats["ratings_sum"] / count, 1)
    print(f"Водій: {driver} | Кількість поїздок: {count} | Виручка: {revenue} грн | Середній рейтинг: {avg_rating}")
```

## Завдання 4. Статуси та бонуси водіїв

На основі `driver_stats` визнач для кожного водія статус і бонус (середній рейтинг береться округленим до 1 знака):
- рейтинг від 4.8 **і** щонайменше 3 поїздки — статус `"Gold"`, бонус 10% від виручки;
- рейтинг від 4.5 — статус `"Silver"`, бонус 5% від виручки;
- рейтинг нижче 4.0 — статус `"Warning"`, бонусу немає;
- в інших випадках — статус `"Standard"`, бонусу немає.

Округли бонус до 2 знаків і додай статус та бонус до статистики водія. Виведи імʼя, статус і бонус кожного водія, а наприкінці — загальну суму бонусів

```python
total_bonuses = 0.0
for driver, stats in driver_stats.items():
    count = stats["rides_count"]
    revenue = stats["total_revenue"]
    avg_reiting = round(stats["ratings_sum"] / count, 1)
    if avg_rating >= 4.8 and count >= 3:
        status = "Gold"
        bonus = revenue * 0.10
    elif avg_rating >= 4.5:
        status = "Silver"
        bonus = revenue * 0.05
    elif avg_rating < 4.0:
        status = "Warning"
        bonus = 0.0
    else:
        status = "Standard"
        bonus = 0.0
    bonus = round(bonus, 2)
    stats["status"] = status
    stats["bonus"] = bonus
    total_bonuses += bonus
    print(f"Водій: {driver} | Статус: {status} | Бонус: {bonus} грн")
total_bonuses = round(total_bonuses, 2)
print(f"\nЗагальна сума бонусів: {total_bonuses} грн")
```

## Завдання 5. Підсумковий звіт за день

Виведи звіт, у якому кожен показник займає окремий рядок:
- загальна виручка за день;
- кількість коректних поїздок і середня вартість поїздки (округли до 2 знаків);
- частка скасованих поїздок від усіх поїздок у журналі, % (округли до 1 знака);
- частка поїздок, оплачених карткою, серед коректних поїздок, % (округли до 1 знака);
- найзавантаженіша година — година, на яку припало найбільше коректних поїздок, і кількість поїздок у цю годину;
- найдовша поїздка: номер, водій і відстань;
- водій з найбільшою виручкою;
- загальна сума бонусів водіям.

Не використовуй функції `max()` і `min()`

```python
total_revenue = round(sum(ride["price"] for ride in valid_rides), 2)
valid_count = len(valid_rides)
avg_ride_price = round(total_revenue / valid_count, 2) if valid_count > 0 else 0.0

cancelled_count = 0
for r in rides:
    if r.get("status") == "cancelled":
        cancelled_count += 1
cancelled_share = round((cancelled_count / len(rides)) * 100, 1) if len(rides) > 0 else 0.0


card_count = 0
for ride in valid_rides:
    if ride.get("payment") == "card":
        card_count += 1
card_share = round((card_count / valid_count) * 100, 1) if valid_count > 0 else 0.0


hour_counts = {}
for ride in valid_rides:
    h = ride["hour"]
    hour_counts[h] = hour_counts.get(h, 0) + 1

busiest_hour = None
max_hour_count = -1
for h, count in hour_counts.items():
    if count > max_hour_count:
        max_hour_count = count
        busiest_hour = h

longest_ride = valid_rides[0]
for ride in valid_rides:
    if ride["distance"] > longest_ride["distance"]:
        longest_ride = ride

top_driver = None
max_driver_revenue = -1.0
for driver, stats in driver_stats.items():
    if stats["total_revenue"] > max_driver_revenue:
        max_driver_revenue = stats["total_revenue"]
        top_driver = driver

total_bonuses_sum = round(sum(stats["bonus"] for stats in driver_stats.values()), 2)

print(f"Загальна виручка за день: {total_revenue} грн")
print(f"Кількість коректних поїздок: {valid_count} | Середня вартість: {avg_ride_price} грн")
print(f"Частка скасованих поїздок: {cancelled_share}%")
print(f"Частка поїздок, оплачених карткою: {card_share}%")
print(f"Найзавантаженіша година: {busiest_hour}:00 (поїздок: {max_hour_count})")
print(f"Найдовша поїздка: №{longest_ride['id']}, водій: {longest_ride['driver']}, відстань: {longest_ride['distance']} км")
print(f"Водій з найбільшою виручкою: {top_driver} ({round(max_driver_revenue, 2)} грн)")
print(f"Загальна сума бонусів водіям: {total_bonuses_sum} грн")
```
